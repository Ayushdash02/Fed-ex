## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# ==========================================================
# 🛠️ Data Wrangling
# Making the Dataset Analysis Ready
# ==========================================================

# Create a copy of the original dataset
df_clean = df.copy()

# ----------------------------------------------------------
# 1. Remove Leading and Trailing Spaces from Column Names
# ----------------------------------------------------------
df_clean.columns = df_clean.columns.str.strip()

# ----------------------------------------------------------
# 2. Remove Leading and Trailing Spaces from Text Columns
# ----------------------------------------------------------
text_columns = df_clean.select_dtypes(include="object").columns

for col in text_columns:
    df_clean[col] = df_clean[col].str.strip()

# ----------------------------------------------------------
# 3. Convert Date Columns to Datetime Format
# ----------------------------------------------------------
date_columns = [
    "Scheduled Delivery Date",
    "Delivered to Client Date",
    "Delivery Recorded Date"
]

for col in date_columns:
    if col in df_clean.columns:
        df_clean[col] = pd.to_datetime(df_clean[col], errors="coerce")

# ----------------------------------------------------------
# 4. Remove Duplicate Records (if any)
# ----------------------------------------------------------
duplicates_before = df_clean.duplicated().sum()

df_clean.drop_duplicates(inplace=True)

duplicates_after = df_clean.duplicated().sum()

# ----------------------------------------------------------
# 5. Check Missing Values
# ----------------------------------------------------------
missing_values = df_clean.isnull().sum().sum()

# ----------------------------------------------------------
# 6. Reset Index
# ----------------------------------------------------------
df_clean.reset_index(drop=True, inplace=True)

# ----------------------------------------------------------
# 7. Display Data Wrangling Summary
# ----------------------------------------------------------
wrangling_summary = pd.DataFrame({
    "Process": [
        "Original Records",
        "Final Records",
        "Duplicates Removed",
        "Remaining Missing Values",
        "Number of Columns"
    ],
    "Result": [
        len(df),
        len(df_clean),
        duplicates_before,
        missing_values,
        df_clean.shape[1]
    ]
})

display(wrangling_summary)

print("✅ Data Wrangling Completed Successfully.")
````

To prepare the dataset for exploratory analysis, several data wrangling steps were performed to improve its quality, consistency, and usability. Column names and text-based fields were standardized by removing leading and trailing whitespace to ensure consistency across categorical variables. Date-related columns were converted into the appropriate datetime format to enable time-based analysis. The dataset was then checked for duplicate records and missing values, and the index was reset after cleaning to maintain a consistent structure.

During this process, the overall quality of the dataset was assessed. #### Chart - 1

````python
# ==========================================================
# 📊 Chart 1 : Top 10 Destination Countries by Shipments
# ==========================================================

# Calculate shipment count by country
top_countries = (
    df.groupby("Country")
      .size()
      .sort_values(ascending=False)
      .head(10)
)

# Create the figure
plt.figure(figsize=(12,6))

# Plot
ax = sns.barplot(
    x=top_countries.values,
    y=top_countries.index
)

# Title and labels
plt.title(
    "Top 10 Destination Countries by Number of Shipments",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Number of Shipments", fontsize=12)
plt.ylabel("Destination Country", fontsize=12)

# Add data labels
for i, value in enumerate(top_countries.values):
    ax.text(value + 5, i, f"{value:,}", va="center", fontsize=10)

# Grid
plt.grid(axis="x", linestyle="--", alpha=0.4)

# Remove top/right borders
sns.despine()

plt.tight_layout()

plt.show()
````

##### 1. Why did you pick the specific chart?

A horizontal bar chart was chosen because it is the most effective visualization for comparing shipment volumes across multiple destination countries. Since the objective is to rank countries based on the number of shipments, the bar chart provides a clear comparison of values and makes it easy to identify the highest and lowest shipment destinations. The horizontal layout also improves readability, especially for country names with longer text, enabling stakeholders to quickly interpret the distribution of shipments across different countries.


##### 2. What is/are the insight(s) found from the chart?

The analysis shows that South Africa received the highest number of shipments (1,406), making it the primary destination in the supply chain network. It is followed by Nigeria (1,194) and Côte d'Ivoire (1,083), indicating that these countries represent the organization's major logistics markets.

The remaining countries, including Uganda, Vietnam, Zambia, Haiti, Mozambique, Zimbabwe, and Tanzania, received comparatively fewer shipments, ranging from 519 to 779 shipments.
````python
# ==========================================================
# 📊 Chart 8 : Shipment Weight vs Freight Cost
# ==========================================================

# Import required library
from scipy.stats import pearsonr

# Create a clean dataset
weight_cost = df[["Weight (Kilograms)", "Freight Cost (USD)"]].copy()

# Convert to numeric
weight_cost["Weight (Kilograms)"] = pd.to_numeric(
    weight_cost["Weight (Kilograms)"],
    errors="coerce"
)

weight_cost["Freight Cost (USD)"] = pd.to_numeric(
    weight_cost["Freight Cost (USD)"],
    errors="coerce"
)

# Remove missing values
weight_cost.dropna(inplace=True)

# Remove extreme outliers (99th percentile)
weight_limit = weight_cost["Weight (Kilograms)"].quantile(0.99)
cost_limit = weight_cost["Freight Cost (USD)"].quantile(0.99)

weight_cost = weight_cost[
    (weight_cost["Weight (Kilograms)"] <= weight_limit) &
    (weight_cost["Freight Cost (USD)"] <= cost_limit)
]

# Calculate correlation
corr, _ = pearsonr(
    weight_cost["Weight (Kilograms)"],
    weight_cost["Freight Cost (USD)"]
)

# Create Figure
plt.figure(figsize=(14,8), dpi=150)

# Scatter Plot
sns.scatterplot(
    data=weight_cost,
    x="Weight (Kilograms)",
    y="Freight Cost (USD)",
    alpha=0.6,
    s=45
)

# Regression Line
sns.regplot(
    data=weight_cost,
    x="Weight (Kilograms)",
    y="Freight Cost (USD)",
    scatter=False,
    color="red",
    line_kws={"linewidth":2}
)

# Title
plt.title(
    f"Shipment Weight vs Freight Cost\n(Pearson Correlation = {corr:.2f})",
    fontsize=18,
    fontweight="bold"
)

# Labels
plt.xlabel("Shipment Weight (Kilograms)", fontsize=13)
plt.ylabel("Freight Cost (USD)", fontsize=13)

# Grid
plt.grid(True, linestyle="--", alpha=0.4)

# Remove extra borders
sns.despine()

# Layout
plt.tight_layout()

# Show Plot
plt.show()
````

The scatter plot reveals the distribution of individual shipments, while the regression line highlights the overall trend. Additionally, the Pearson correlation coefficient quantifies the strength and direction of the relationship, making it easier to assess whether freight cost increases as shipment weight increases.

##### 2. What is/are the insight(s) found from the chart?

#### Chart - 14 - Correlation Heatmap

````python
# ==========================================================
# 📊 Chart 14 : Correlation Heatmap of Logistics Variables
# ==========================================================

# Select important numerical logistics variables
heatmap_df = df[
    [
        "Weight (Kilograms)",
        "Freight Cost (USD)",
        "Line Item Quantity",
        "Line Item Value",
        "Pack Price",
        "Unit Price",
        "Line Item Insurance (USD)"
    ]
].copy()

# Convert all columns to numeric
for col in heatmap_df.columns:
    heatmap_df[col] = pd.to_numeric(
        heatmap_df[col],
        errors="coerce"
    )

# Remove missing values
heatmap_df.dropna(inplace=True)

# Calculate correlation matrix
corr_matrix = heatmap_df.corr(method="pearson")

# Create Figure
plt.figure(figsize=(12, 8), dpi=150)

# Heatmap
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True,
    vmin=-1,
    vmax=1,
    cbar_kws={"label": "Correlation Coefficient"}
)

# Title
plt.title(
    "Correlation Heatmap of Key Logistics Variables",
    fontsize=18,
    fontweight="bold",
    pad=15
)

# Rotate labels
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

# Improve spacing
plt.tight_layout()

# Display
plt.show()
````

##### 1. Why did you pick the specific chart?

A Correlation Heatmap was selected because it is one of the most effective visualization techniques for identifying the strength and direction of relationships among multiple numerical variables simultaneously. Unlike scatter plots, which compare only two variables, the heatmap provides a comprehensive overview of all pairwise correlations in a single visualization. This enables analysts to quickly identify variables that move together, detect multicollinearity, and understand the key factors influencing shipment costs, insurance, and item values.

##### 2. What is/are the insight(s) found from the chart?

The heatmap reveals several important relationships among the logistics variables:

*    Line Item Value and Line Item Insurance (USD) show a very strong positive correlation (0.96), indicating that shipments with higher item values generally require higher insurance coverage.
*    Line Item Quantity and Line Item Value have a strong positive correlation (0.82). This suggests that shipments containing more units usually result in a higher total shipment value.
*    Line Item Quantity and Line Item Insurance (USD) exhibit a strong positive correlation (0.78), implying that larger orders often require increased insurance protection.
*    Weight (Kilograms) and Line Item Quantity have a strong positive relationship (0.71), showing that shipment weight generally increases as the number of items increases.
*    Weight (Kilograms) and Line Item Value also demonstrate a strong positive correlation (0.69), indicating that heavier shipments often contain goods of higher value.
*    Weight (Kilograms) and Freight Cost (USD) have a moderate positive correlation (0.47), meaning shipment weight influences transportation cost, although other operational factors such as shipping mode, destination, and route also contribute.
*    Pack Price and Unit Price display only weak relationships with most logistics variables, suggesting that pricing is relatively independent of shipment volume, insurance, or freight cost.
*    Most negative correlations are very close to zero, indicating no significant inverse relationships among the selected variables.
